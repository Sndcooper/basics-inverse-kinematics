```python
from math import *
armLength = 2
angle1 = -pi/4
angle2 = pi/8
arm1X = armLength*cos(angle1)
arm1Y = armLength*sin(angle1)
arm2X = arm1X + armLength*cos(angle2)
arm2Y = arm1Y + armLength*sin(angle2)

print(arm1X, arm1Y)
print(arm2X, arm2Y)

x = np.array([0, arm1X, arm2X])
y = np.array([0, arm1Y, arm2Y])

plt.plot(x, y, marker='o', markersize=10)
plt.xlabel("x")
plt.ylabel("y")
plt.title("forward kinematics")

plt.xticks([-4, -3, -2, -1, 0, 1, 2, 3, 4])
plt.yticks([-4, -3, -2, -1, 0, 1, 2, 3, 4])

plt.legend(["arm"])

plt.plot()
```

```python
from math import *

X, Y = 0,0

L = 2

r = sqrt(X**2 + Y**2)
cosA2 = r**2 - 2*L*L
cosA2 /= L*L*2
angle2 = acos(cosA2) #elbow angle negative if u want it upside

phi = atan2(Y,X) #angle bw x-axis and line from origin to (X,Y)
psi = atan2(L*sin(angle2), L+L*cos(angle2)) #angle bw line from origin to (X,Y) and line from origin to arm1

angle1 = phi - psi #shoulder angle

arm1X = L*cos(angle1)
arm1Y = L*sin(angle1)
arm2X = arm1X + L*cos(angle2)
arm2Y = arm1Y + L*sin(angle2)

x = np.array([0, arm1X, arm2X])
y = np.array([0, arm1Y, arm2Y])

plt.plot(x, y, marker='o', markersize=10)
plt.xlabel("x")
plt.ylabel("y")
plt.title("inverse kinematics")

plt.xticks([-4, -3, -2, -1, 0, 1, 2, 3, 4])
plt.yticks([-4, -3, -2, -1, 0, 1, 2, 3, 4])

plt.legend(["arm"])

plt.show()
```
